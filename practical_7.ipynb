Implementation of making a chance problem using dynamic programming


```python
import time


def making_change(coins, amount):
    # dp[i] = minimum number of coins needed
    # to make amount i

    dp = [float('inf')] * (amount + 1)

    # 0 coins are needed to make amount 0
    dp[0] = 0

    # Calculate minimum coins for every amount
    for i in range(1, amount + 1):

        for coin in coins:

            if coin <= i:
                dp[i] = min(
                    dp[i],
                    dp[i - coin] + 1
                )

    return dp[amount]


# --------------------------------
# User Input
# --------------------------------

n = int(input("Enter number of coin denominations: "))

coins = []

for i in range(n):
    coin = int(input(f"Enter value of coin {i + 1}: "))
    coins.append(coin)

amount = int(input("Enter amount to make: "))


# --------------------------------
# Measure Execution Time
# --------------------------------

start_time = time.perf_counter()

result = making_change(coins, amount)

end_time = time.perf_counter()

execution_time = end_time - start_time


# --------------------------------
# Output
# --------------------------------

if result == float('inf'):
    print("\nChange cannot be made for the given amount.")
else:
    print("\nMinimum number of coins:", result)

print("Execution time:", execution_time, "seconds")

```

READ ME

The Making Change problem is efficiently solved using Dynamic Programming by finding the minimum number of coins needed for a given amount.\
 The algorithm has a **time complexity of O(n × A)** and a **space complexity of O(A)**.\
 The execution time is measured using `time.perf_counter()` and may vary depending on the system.
