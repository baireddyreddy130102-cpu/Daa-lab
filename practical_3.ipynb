maximum heapsort

```python
import time

n = int(input("Enter number of elements: "))

arr = []
for i in range(n):
    arr.append(int(input(f"Enter element {i + 1}: ")))

start = time.perf_counter()

maximum = arr[0]

for i in range(1, n):
    if arr[i] > maximum:
        maximum = arr[i]

end = time.perf_counter()

print("Maximum element:", maximum)
print("Execution time:", end - start, "seconds")

print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

Max Heap Sort arranges elements using a max heap, repeatedly placing the largest element at the end.
Min Heap Sort uses a min heap, repeatedly placing the smallest element first; both have O(n log n) time complexity.
