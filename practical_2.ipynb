linear search

```python
# Linear Search with user input and time complexity

import time

n = int(input("Enter number of elements: "))
arr = []

for i in range(n):
    arr.append(int(input(f"Enter element {i + 1}: ")))

key = int(input("Enter element to search: "))

start = time.perf_counter()

found = False
for i in range(n):
    if arr[i] == key:
        found = True
        position = i
        break

end = time.perf_counter()

if found:
    print("Element found at position:", position + 1)
else:
    print("Element not found")

print("Execution time:", end - start, "seconds")
print("Time Complexity: O(n)")
print("Best Case: O(1)")
print("Worst Case: O(n)")
print("Space Complexity: O(1)")
```

binary search

```python
import time

# User input
n = int(input("Enter number of elements: "))

arr = []
for i in range(n):
    arr.append(int(input(f"Enter element {i + 1}: ")))

# Binary search requires sorted data
arr.sort()

key = int(input("Enter element to search: "))

# Start time
start = time.perf_counter()

low = 0
high = n - 1
found = False

while low <= high:
    mid = (low + high) // 2

    if arr[mid] == key:
        found = True
        position = mid
        break
    elif arr[mid] < key:
        low = mid + 1
    else:
        high = mid - 1

# End time
end = time.perf_counter()

if found:
    print("Element found at position:", position + 1)
else:
    print("Element not found")

print("Execution time:", end - start, "seconds")
print("Best Case Time Complexity: O(1)")
print("Worst Case Time Complexity: O(log n)")
print("Average Case Time Complexity: O(log n)")
print("Space Complexity: O(1)")
```
