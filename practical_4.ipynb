iteraive function

```python
import time

def iterative_sum(n):
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# User input
n = int(input("Enter a number: "))

# Measure execution time
start = time.perf_counter()

result = iterative_sum(n)

end = time.perf_counter()

print("Sum:", result)
print("Execution time:", end - start, "seconds")
print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

recursive function

```python
import time

def recursive_sum(n):
    if n == 0:
        return 0
    return n + recursive_sum(n - 1)

# User input
n = int(input("Enter a positive number: "))

# Measure execution time
start = time.perf_counter()

result = recursive_sum(n)

end = time.perf_counter()

print("Sum:", result)
print("Execution time:", end - start, "seconds")
print("Time Complexity: O(n)")
print("Space Complexity: O(n)")
```

read me

An iterative function uses loops to repeat a task and usually has O(n) time and O(1) space complexity.
A recursive function calls itself repeatedly until a base condition is reached, with O(n) time and O(n) space complexity for this example.
